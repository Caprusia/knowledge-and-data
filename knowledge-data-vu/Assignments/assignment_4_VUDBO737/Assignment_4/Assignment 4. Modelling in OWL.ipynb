### Task 5a (10 Points): Property characteristics

Add at least **four** different OWL property characteristics.

Examples are
- transitivity
- symmetricity
- functionality
- (ir)reflexivity
- property chain
- disjoint properties
- etc

All four different characteristics can be asserted on one property (or on four different properties).

List and describe the four (or more) property characteristics you defined (i.e. axioms) in the textfield below. Do not forget to motivate your choices.

1. Functional property: climb:hasName. Every individual has at most one name. This avoids assigning two different names to the same climbing entity.

2. Irreflexive property: climb:locatedIn. A climbing route, equipment item, or other entity cannot be located in itself. This prevents self-location statements.

3. Symmetric property: climb:hasClimbingPartner. If one climber is recorded as a climbing partner of another climber, the reverse partnership also holds. This reflects the mutual nature of a climbing partnership.

4. Transitive property: climb:isPartOf. If one entity is part of a second entity, and the second entity is part of a third entity, then the first entity is part of the third entity. This can represent routes being part of areas that are themselves part of larger regions.
